```python
from transformers import AutoTokenizer, AutoModel
import torch
import torch.nn.functional as F

def mean_pooling(model_output, attention_mask):
    token_embeddings = model_output[0]  # First element of model_output contains all token embeddings
    input_mask_expanded = attention_mask.unsqueeze(-1).expand(token_embeddings.size()).float()
    sum_embeddings = torch.sum(token_embeddings * input_mask_expanded, 1)
    sum_mask = torch.clamp(input_mask_expanded.sum(1), min=1e-9)
    return sum_embeddings / sum_mask

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

sentences = ["Krishna is a Hindu deity.", "Yamunaji is a sacred river in India."]

tokenized_input = tokenizer(sentences, padding=True, truncation=True, return_tensors="pt")
with torch.no_grad():
    model_output = model(**tokenized_input)


sentence_embeddings = mean_pooling(model_output, tokenized_input['attention_mask'])
sentence_embeddings = F.normalize(sentence_embeddings, p=2, dim=1)

print("Sentence embeddings:")
for i, sentence in enumerate(sentences):
    print(f"Sentence: {sentence}")
    print(f"Embedding: {sentence_embeddings[i].numpy()}\n")
```

```python
import numpy as np
import faiss

#Step 1: Create random vectors to simulate your embeddings
dimension = 128
num_vectors = 100
vectors = np.random.rand(num_vectors, dimension).astype('float32')

#Step 2: Create a FAISS index
index = faiss.IndexFlatL2(dimension)  # Using L2 distance for similarity

#Step 3: Add vectors to the index
index.add(vectors)

#Step 4: Query the index with a new vector
query_vector = np.random.rand(1, dimension).astype('float32')
k = 5  # Number of nearest neighbors to retrieve
distances, indices = index.search(query_vector, k)

print("Query vector:", query_vector)
print("Nearest neighbors' indices:", indices)
print("Nearest neighbors' distances:", distances)
```

```python
from azure.core.credentials import AzureKeyCredential
from azure.ai.formrecognizer import DocumentAnalysisClient
from langchain_text_splitters import RecursiveCharacterTextSplitter
from dotenv import load_dotenv
import os

# Load environment variables
load_dotenv()

endpoint = os.getenv("AZURE_DOCUMENT_INTELLIGENCE_ENDPOINT")
key = os.getenv("AZURE_DOCUMENT_INTELLIGENCE_KEY")

# Check that they loaded
print("Endpoint:", endpoint)
print("Key loaded:", key is not None)

# Initialize Document Analysis Client
document_analysis_client = DocumentAnalysisClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(key)
)

doc_path = "C:/Users/Admin/Desktop/Azure-OpenAI/202201487_Resume_v1.pdf"

# Analyze document
with open(doc_path, "rb") as f:
    poller = document_analysis_client.begin_analyze_document(
        "prebuilt-document",
        document=f
    )

result = poller.result()

# Extract text
extracted_text = result.content

print("Extracted text:")
print(extracted_text)

# Split text into chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = splitter.split_text(extracted_text)

# Print chunks
print(f"\nNumber of chunks: {len(chunks)}\n")

for i, chunk in enumerate(chunks):
    print(f"Chunk {i + 1}:\n{chunk}\n")
```
